```python
import examples.transformations.youtube
import networkx as nx

from eltstar.transformation import manager
```

```python
lineage = manager.lineage
```

```python
from eltstar.graph import TableNode


g=lineage.graph.copy()

nodes = list(g.nodes)

for node in nodes:
    if not isinstance(node, TableNode):
        continue
    in_edges = g.in_edges(node,keys=True)
    out_edges = g.out_edges(node,keys=True)
    if in_edges and out_edges:
        for in_edge in in_edges:
            for out_edge in out_edges:
                g.add_edge(in_edge[0], out_edge[1], key=in_edge[2])
        g.remove_node(node)

lineage.draw(g)
```
